## **Type conversion**


Los tipos de conversión simplemente significa transformar datos de un tipo a otro. Tomemos un ejemplo sencillo:

```python
# number = input("Enter a number: ")
# number = number*3
# number
# print(number)
```

El resultado que obtendrás es 555. La razón de esto es que la instrucción de entrada toma un valor en formato de cadena, y al usar *3 se repetirá el valor de la cadena 3 veces. Sin embargo, si queremos realizar una operación matemática con la variable numérica, necesitamos convertirla a un entero de la siguiente manera:

```python
# number = input('Enter a number: ' )
# number = int(number) *3
# print(number)
```

En el mismo programa, realizamos cambios menores al convertir la variable de número de cadena a entero, y el resultado que obtuvimos fue 15. Esa conversión de datos de cadena a entero se llama conversión de tipos.

Python puede realizar la conversión de tipos de dos maneras: implícita y explícita.

## **Implicit type casting**


Cuando Python convierte automáticamente el tipo de dato de un valor, se trata de un casting de tipos implícito. También se le conoce como coerción, donde el intérprete convierte un tipo de dato en otro sin que el programador tenga que dar instrucciones explícitas. Esto suele ocurrir cuando se usan operandos de diferentes tipos en una expresión, y Python los convierte a un tipo común para evitar pérdida de datos o comportamientos inesperados.


```python
# n = 52
# o = 2.0
# prod = n*o
```

```python
# prod
```

```python
# type(prod)
```

En el casting implícito, el tipo de dato de la variable prod se convierte automáticamente a float cuando se calcula el producto de un entero ***n*** y un float ***o***.

En los tipos de conversión explícita, el tipo de dato de una variable se cambia deliberadamente por el programador. También se conoce como 'casting' y se realiza manualmente mediante funciones integradas como int(), float(), str(), bool(), etc. Consulta el siguiente ejemplo para una mejor comprensión:

## **Nested if-elif-else statement**


Cuando usamos una estructura if-elif-else dentro de otra estructura if-else, esto se llama una sentencia anidada if-elif-else. Permite verificar múltiples expresiones de prueba y ejecutar diferentes bloques de código para más de dos condiciones.


```python
# Estrutura de sentencia anidada if-elif-else.

if 'A' == 'T':
    print('Adenina y timina son iguales')

else:
    if 'A' > 'T':
        print('Adenina es mayor que timina')

    else:
        print('Timina es mayor que adenina')
```

El ejemplo anterior demuestra el uso de if anidados. El ejemplo comparará los valores de dos variables, A y T; si son iguales, se imprimirá el mensaje 'A y T son iguales'. De lo contrario, en la parte del else, se ejecuta otro if para realizar verificaciones adicionales.

```python
# Secuencia original de ADN (Salmo salar)
secuencia_marina = "AAGCTTCGGCGGCAACCACCATAGCCATCGTC"

print("Clasificación detallada de la secuencia marina:\n")

for nucleotido in secuencia_marina:
    # --- PRIMER NIVEL (Grupo Químico) ---
    if nucleotido == "A" or nucleotido == "G":
        # --- SEGUNDO NIVEL (Anidado para Purinas) ---
        if nucleotido == "A":
            print(f"Base: {nucleotido} -> Es una Purina (Adenina)")
        else:
            print(f"Base: {nucleotido} -> Es una Purina (Guanina)")

    elif nucleotido == "C" or nucleotido == "T":
        # --- SEGUNDO NIVEL (Anidado para Pirimidinas) ---
        if nucleotido == "C":
            print(f"Base: {nucleotido} -> Es una Pirimidina (Citosina)")
        else:
            print(f"Base: {nucleotido} -> Es una Pirimidina (Timina)")

    else:
        # Por si existe algún error de lectura, un espacio o un nucleotido indeterminado ('N')
        print(f"Base: {nucleotido} -> Carácter no reconocido")
```

Cuando necesitas repetir la ejecución de unos cuantos conjuntos de instrucciones, usamos iteraciones o bucles. Una iteración simplemente significa repetir el conjunto de instrucciones hasta que alguna condición sea verdadera o falsa, o hasta que se haya alcanzado el número de repeticiones deseadas.


Tomemos un ejemplo muy sencillo: si quieres imprimir una secuencia cinco veces, en ese caso puedes escribir print(“AAATTTGCGC”) cinco veces o bien usar una sentencia iterativa. La mejor opción es usar una sentencia iterativa. En Python se utilizan dos tipos de sentencias iterativas.

En un ciclo for, sabemos cuántas veces se ejecutará el ciclo.

### **Pass statement**

Esta declaración se usa cuando se requiere una instrucción en particular pero no quieres ejecutar ese comando. Es como una declaración nula. No pasará nada cuando se ejecute.

Sintaxis:

pass

A veces planificamos una función o un ciclo, pero la parte de implementación la queremos escribir en el futuro. En ese caso, podemos colocar sentencias 'pass' para que el intérprete no genere ningún problema en tiempo de ejecución.
