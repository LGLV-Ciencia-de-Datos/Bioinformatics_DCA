En la lección anterior realizamos una consulta BLAST sobre nuestra proteína de COVID.

También podemos recuperar el mismo archivo estructural de SARS-CoV-2 desde otra base de datos: el **PDB (Protein Data Bank)**. La base de datos PDB almacena registros de proteínas que contienen información de coordenadas de cada átomo, la cual usaremos para visualizar la proteína de SARS-CoV-2.

```python
# ID de la proteína que estamos buscando (copiado de la clase del Día 7)
seq_id = "pdb|6YYT|A"
```

```python
# Extraemos el ID para poder descargar el archivo PDB desde la base de datos Protein Data Bank
id = seq_id.split("|")[1]
```

```python
id
```

El formato de archivo **Protein Data Bank (PDB)** es un formato de texto que describe las estructuras tridimensionales de moléculas almacenadas en el Protein Data Bank.

Descargamos el archivo PDB con `requests` (compatible con Colab):

```python
import requests

url = f"https://files.rcsb.org/download/{id}.pdb"
response = requests.get(url)

if response.status_code == 200:
    with open(f"{id}.pdb", "w") as file:
        file.write(response.text)
    print(f"¡Archivo {id}.pdb descargado exitosamente!")
else:
    print(f"Error al descargar. Código de estado: {response.status_code}")
```

### Lectura de archivos PDB con Biopython

`Bio.PDB` es un módulo de Biopython que se enfoca en trabajar con estructuras cristalinas de macromoléculas biológicas. Entre otras cosas, `Bio.PDB` incluye una clase `PDBParser` que produce un objeto `Structure`, el cual puede usarse para acceder a los datos atómicos del archivo de manera conveniente.

Más sobre esto en un video posterior :)

```python
from Bio.PDB import PDBParser  # PDBParser - analizador de archivos PDB
```

```python
parser = PDBParser()
# Después de parsear, podemos obtener la estructura de la proteína usando get_structure
structure = parser.get_structure('6YYT', '6YYT.pdb')
structure
```

#### Identificar el número de cadenas

Para identificar cuántas cadenas tiene esta proteína viral 6YYT de COVID, usamos `chain.id`, que nos da la lista de las cadenas presentes.

```python
for chain in structure[0]:
    print(f'ID de la cadena: {chain.id}')
```

Vemos que esta polimerasa viral de SARS-CoV-2 tiene **8 cadenas** o 8 proteínas accesorias, representadas con una sola letra.

Así es como se ve la proteína 6YYT de SARS-CoV-2:

- Dos hebras helicoidales con diferentes tonos de azul son la hebra molde de ARN y su hebra producto.
- El grueso de las cintas rojas es la **polimerasa**, una enzima (proteína funcional) que hace copias de la cadena de ARN. Esta polimerasa es un objetivo atractivo para los antivirales y vacunas de COVID-19.
- Si giramos la molécula, podemos ver las cintas amarillas y naranjas, que son las proteínas virales que ayudan a la polimerasa a mantenerse en el camino y copiar largas porciones de la cadena de ARN.
