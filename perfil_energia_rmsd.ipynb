```python
import matplotlib.pyplot as plt

# 1. Datos obtenidos de tu AD vina
modos = [1, 2, 3, 4, 5, 6, 7, 8, 9]
afinidades = [-8.4, -7.9, -7.6, -7.2, -6.8, -6.5, -6.1, -5.9, -5.5]
rmsd_lower = [0.85, 1.42, 2.15, 3.80, 4.12, 5.30, 6.45, 7.10, 8.25]  # Usado para la serie RMSD

# 2. Configurar la figura y los ejes
fig, ax1 = plt.subplots(figsize=(9, 6), dpi=150)

# Paleta de colores
color_afinidad = '#1b365d'  # Azul oscuro
color_rmsd = '#c05c53'      # Rojo/Coral
color_umbral = '#6fa86b'    # Verde

# --- EJE IZQUIERDO: Afinidad de Unión ---
line1 = ax1.plot(modos, afinidades, color=color_afinidad, marker='o', linewidth=2.5,
                 markersize=8, label='Afinidad Vina (kcal/mol)')
ax1.set_xlabel('Modo / Pose de Docking', fontsize=12, fontweight='bold', labelpad=10)
ax1.set_ylabel('Afinidad de Unión (kcal/mol)', color=color_afinidad, fontsize=12, fontweight='bold', labelpad=10)
ax1.tick_params(axis='y', labelcolor=color_afinidad, labelsize=11)
ax1.tick_params(axis='x', labelsize=11)
ax1.set_xticks(modos)
ax1.set_ylim(-8.6, -5.3)

# Grid de fondo
ax1.grid(True, linestyle='--', alpha=0.5, color='#cccccc')

# --- EJE DERECHO: RMSD ---
ax2 = ax1.twinx()
line2 = ax2.plot(modos, rmsd_lower, color=color_rmsd, marker='s', linestyle='--', linewidth=2.5,
                 markersize=7, label='RMSD (Å)')
ax2.set_ylabel('RMSD vs Cristalino (Å)', color=color_rmsd, fontsize=12, fontweight='bold', labelpad=10)
ax2.tick_params(axis='y', labelcolor=color_rmsd, labelsize=11)
ax2.set_ylim(0, 8.8)

# --- LÍNEA DE UMBRAL DE VALIDACIÓN ---
line3 = ax2.axhline(y=2.0, color=color_umbral, linestyle=':', linewidth=2, label='Umbral Validación (2.0 Å)')

# --- TÍTULO ---
plt.title('Perfil de Energía de Unión y Desviación RMSD (Redocking PDB: 9JQK)',
          fontsize=13, fontweight='bold', pad=15)

# --- LEYENDA (Combinada para ambos ejes) ---
lines = line1 + line2 + [line3]
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='lower center', bbox_to_anchor=(0.5, -0.22),
           ncol=3, frameon=True, facecolor='white', edgecolor='#cccccc', fontsize=10)

# Ajuste de diseño final
plt.tight_layout()

# Mostrar o guardar el gráfico
plt.show()
# plt.savefig("grafico_docking.png", dpi=300, bbox_inches='tight')
```

Características implementadas:
* Eje Y doble (ax.twinx()): Permite representar en una misma gráfica la Afinidad de Unión (escala negativa a la izquierda) y el RMSD (escala positiva a la derecha).

* Marcadores y estilos: Círculos con línea continua azul para la afinidad, y cuadrados con línea punteada roja para el RMSD.

* Línea de referencia verde: Creada con axhline(y=2.0) para marcar el umbral de validación a 2.0 Å.

* Leyenda horizontal inferior: Posicionada debajo del eje X combinando las series de ambos ejes.
