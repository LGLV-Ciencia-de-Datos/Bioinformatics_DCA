### Paso B: Cargar, añadir hidrógenos, generar 3D y minimizar

Nota importante sobre Paritaprevir: Al ser una molécula grande y rígida por el macrociclo, usaremos el método ETKDGv3 de RDKit con restricciones de estereoquímica y el campo de fuerzas MMFF94s.

```python
import py3Dmol
from rdkit import Chem
from rdkit.Chem import AllChem

# 1. Cargar la molécula desde tu archivo SDF (o usar SMILES explícito con estereoquímica)
# Reemplaza 'paritaprevir.sdf' por la ruta de tu archivo
suppl = Chem.SDMolSupplier('Conformer3D_COMPOUND_CID_13817.sdf')
mol = next(suppl)

# Alternativa directa por SMILES (si el SDF tuviera problemas de lectura):
# smiles = "CC1(C)C2CC2C(=O)NC3(C(=O)NS(=O)(=O)C4(CC4)C5=NC(=CS5)C6=CC=C(C=C6)N7C=CC=N7)CCCCCC(=O)NC8CCCC18"
# mol = Chem.MolFromSmiles(smiles)

# 2. Agregar hidrógenos explícitos (esencial para 3D y minimización)
mol_3d = Chem.AddHs(mol)

# 3. Generar coordenadas 3D preservando la estereoquímica (ETKDGv3)
params = AllChem.ETKDGv3()
params.randomSeed = 42
params.pruneRmsThresh = 0.5
AllChem.EmbedMolecule(mol_3d, params)

# 4. Minimización de energía con campo de fuerzas MMFF94s
# Se ejecutan hasta 500 iteraciones para asegurar convergencia
AllChem.MMFFOptimizeMolecule(mol_3d, mmffVariant='MMFF94s', maxIters=500)

# 5. Guardar resultado limpio en formato MOL2 o PDB
Chem.MolToMolFile(mol_3d, 'paritaprevir_3d_minimizada.mol')
Chem.MolToPDBFile(mol_3d, 'paritaprevir_3d_minimizada.pdb')

print("¡Generación 3D y minimización completadas con éxito!")
```
