Aquí mostraremos algunos ejemplos más de análisis de secuencias, usando el módulo `Bio.SeqIO` descrito en una de las clases anteriores, junto con la interfaz de gráficos `matplotlib` de Python.

### Obtener los datos

```python
# Descargar el archivo de ejemplo de orquídeas (compatible con Colab)
import urllib.request

url = "https://raw.githubusercontent.com/biopython/biopython/master/Doc/examples/ls_orchid.fasta"
urllib.request.urlretrieve(url, "ls_orchid.fasta")
print("Archivo ls_orchid.fasta descargado exitosamente.")
```

```python
!pip install biopython matplotlib
```

### Parsear el archivo FASTA

```python
from Bio import SeqIO

sequences = []
for seq_record in SeqIO.parse("ls_orchid.fasta", "fasta"):
    sequences.append(seq_record.seq)

print(f"Se leyeron {len(sequences)} secuencias.")
```

```python
sequences[:5]  # primeras 5 secuencias
```

### Graficar la longitud de las secuencias

```python
import matplotlib.pyplot as plt

lengths = [len(seq) for seq in sequences]

plt.hist(lengths, bins=20, edgecolor='black')
plt.title(f"Distribución de longitudes\n({len(lengths)} secuencias de orquídea)")
plt.xlabel("Longitud de la secuencia (pb)")
plt.ylabel("Frecuencia")
plt.show()
```

### Calcular y graficar el contenido de GC

```python
from Bio.SeqUtils import gc_fraction

# gc_fraction devuelve un valor entre 0 y 1; lo convertimos a porcentaje
gc_values = sorted([gc_fraction(seq) * 100 for seq in sequences])

print(f"Contenido de GC mínimo: {min(gc_values):.2f}%")
print(f"Contenido de GC máximo: {max(gc_values):.2f}%")
```

```python
plt.plot(gc_values)
plt.title(f"{len(gc_values)} secuencias de orquídea\nGC% de {min(gc_values):.1f} a {max(gc_values):.1f}")
plt.xlabel("Secuencias (ordenadas por GC%)")
plt.ylabel("GC%")
plt.show()
```

```python
plt.hist(gc_values, bins=20, edgecolor='black', color='teal')
plt.title("Distribución del contenido de GC")
plt.xlabel("GC%")
plt.ylabel("Frecuencia")
plt.show()
```
