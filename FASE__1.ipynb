```python
# ============================================================
# FASE 1: Carga y control de calidad de la secuencia problema
# ============================================================
# Objetivo: Verificar que la secuencia está bien formateada,
# calcular longitud, composición de bases y GC%, y visualizar
# un resumen rápido antes de cualquier búsqueda.

!pip install -q biopython

from Bio import SeqIO
from Bio.SeqUtils import gc_fraction
from collections import Counter
import matplotlib.pyplot as plt

# --- 1. Sube tu archivo FASTA a Colab (o pégalo aquí) ---
# Opción A: Subir archivo
from google.colab import files
uploaded = files.upload()          # selecciona tu .fasta
fasta_file = list(uploaded.keys())[0]

# Opción B: Pegar la secuencia directamente (descomenta si prefieres)
# fasta_text = """>secuencia_problema
# # ATGCGT..."""
# with open("problema.fasta", "w") as f:
#      f.write(fasta_text)
# fasta_file = "problema.fasta"

# --- 2. Leer la secuencia ---
record = SeqIO.read(fasta_file, "fasta")
seq = record.seq
print(f"ID: {record.id}")
print(f"Descripción: {record.description}")
print(f"Longitud: {len(seq)} nt")
print(f"GC%: {gc_fraction(seq)*100:.2f}%")

# --- 3. Composición de bases ---
counts = Counter(str(seq).upper())
print("\nComposición de bases:")
for base in "ACGTN":
    print(f"  {base}: {counts.get(base, 0)}")

# --- 4. Gráfico simple de composición ---
bases = ["A", "C", "G", "T"]
vals = [counts.get(b, 0) for b in bases]
plt.bar(bases, vals)
plt.title("Composición de nucleótidos")
plt.ylabel("Conteo")
plt.show()

print("\n✅ Fase 1 completada. Si la secuencia tiene N's o es muy corta (<200 nt) revisa la calidad.")
```
