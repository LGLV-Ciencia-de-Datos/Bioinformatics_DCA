Un **motivo de secuencia** (*sequence motif*) es un patrón de secuencia de nucleótidos o aminoácidos. Los motivos de secuencia se forman por la disposición tridimensional de aminoácidos que pueden no ser adyacentes. Biopython proporciona un módulo separado, `Bio.motifs`, para acceder a las funcionalidades de motivos de secuencia.

Los objetos Motif en Biopython (pertenecientes al módulo Bio.motifs) sirven para representar, analizar y buscar patrones biológicos conservados en secuencias de ADN, ARN o proteínas. Un motivo es un segmento de secuencia corto y recurrente que tiene un significado funcional o estructural específico (por ejemplo, el sitio de unión de un factor de transcripción).

¿Para qué sirven exactamente?

El módulo Bio.motifs encapsula toda la información matemática y biológica del patrón en un solo objeto para realizar las siguientes tareas:

* Representación probabilística: Convierte un conjunto de secuencias alineadas en una matriz de conteo de frecuencias (PFM), matriz de probabilidad (PWM) o matriz de peso posicional (PSSM).

* Cálculo de consenso: Genera la secuencia consenso (la versión más representativa del patrón) y su secuencia degenerada (usando el código IUPAC).

* Búsqueda y escaneo: Permite escanear genomas o secuencias largas para localizar dónde se encuentra presente ese patrón y evaluar qué tan fuerte es la coincidencia mediante puntuaciones de log-odds.

* Generación de gráficos: Permite crear Sequence Logos (diagramas donde la altura de cada letra indica qué tan conservado está ese nucleótido/aminoácido en esa posición).

* Lectura y escritura de formatos: Importa y exporta patrones desde formatos estándar de bioinformática como JASPAR, MEME, TRANSFAC o TRANSFAC/BLOCKS.

### Crear un motivo de ADN simple

Vamos a crear una secuencia de motivo de ADN:

```python
from Bio.Seq import Seq

instances = [
    Seq("TACAAAATGATGATGTAGTAGTA"),
    Seq("TACGCAGATTAGTAGTAGTAGTA"),
    Seq("TACACATAGTATCGTGCTAGTCA"),
    Seq("TACCCATATAGCTAGCTGATCGA"),
    Seq("AACCCAGTAATGATGTATATAGT"),
    Seq("AATGCATGATTTAGATGATATGT"),
    Seq("AATGCAGAGTCTACTGTACTACG")
]
```

```python
m = motifs.create(instances)
```

```python
print(m)
```

Las instancias se guardan en un atributo **`.instances`**

```python
m.alignment.sequences
```

El objeto Motif tiene un atributo **`.counts`** que contiene los conteos de cada nucleótido en cada posición. Al imprimir esta matriz de conteos se muestra en un formato fácilmente legible:

```python
print(m.counts)
```

Podemos acceder a estos conteos como un diccionario:

```python
m.counts['A'][0:5]
```

También podemos acceder directamente a las columnas de la matriz de conteos:

```python
m.counts[:, 3]
```

El motivo tiene una secuencia **consensus** (*consenso*) asociada, definida como la secuencia de letras a lo largo de las posiciones del motivo para las cuales se obtiene el valor más grande en las columnas correspondientes de la matriz `.counts`:

```python
m.consensus
```

Así como una secuencia **anticonsensus** (*anticonsenso*), correspondiente a los valores más pequeños en las columnas de la matriz `.counts`:

```python
m.anticonsensus
```

Ten en cuenta que hay cierta ambigüedad en la definición de la secuencia consenso y anticonsenso si en algunas columnas varios nucleótidos tienen el conteo máximo o mínimo.

También puedes pedir una secuencia consenso **degenerada**, en la que se usan nucleótidos ambiguos para las posiciones donde hay múltiples nucleótidos con conteos altos:

```python
m.degenerate_consensus
```

En Biopython, el módulo Bio.motifs está diseñado para representar, manipular y analizar motivos de secuencia (patrones o firmas de nucleótidos o aminoácidos que tienen un significado biológico conservado, como los sitios de unión de factores de transcripción).
### **1. Conceptos fundamentales del objeto Motif**

Un motivo en Biopython se construye habitualmente a partir de un conjunto de secuencias alineadas de la misma longitud y consta de tres niveles de representación:

    Instancias (instances): Las secuencias individuales a partir de las cuales se infiere el motivo.

    Matriz de Frecuencias de Posición (PFM - Position Frequency Matrix): Cuenta cuántas veces aparece cada monómero (A, C, G, T) en cada posición del alineamiento.

    Matriz de Probabilidad de Posición (PPM - Position Probability Matrix): La frecuencia relativa de cada nucleótido en cada posición (los valores de cada columna suman 1).

    Matriz de Peso de Posición (PWM / PSSM - Position-Specific Scoring Matrix): Representa el modelo estadístico que calcula la puntuación (score) de log-odds de encontrar cada base en comparación con un modelo de fondo (background).

### **2. Creación básica de un objeto Motif**

Se pueden crear motivos manualmente pasando una lista de objetos Seq:

```python
from Bio import motifs
from Bio.Seq import Seq

# 1. Definir un conjunto de secuencias asociadas al motivo
# instances = [
#     Seq("TACGAT"),
#     Seq("TATAAT"),
#     Seq("TACAAT"),
#     Seq("GATACT"),
#     Seq("TATGAT"),
# ]

instances = [
    Seq("TACAAAATGATGATGTAGTAGTA"),
    Seq("TACGCAGATTAGTAGTAGTAGTA"),
    Seq("TACACATAGTATCGTGCTAGTCA"),
    Seq("TACCCATATAGCTAGCTGATCGA"),
    Seq("AACCCAGTAATGATGTATATAGT"),
    Seq("AATGCATGATTTAGATGATATGT"),
    Seq("AATGCAGAGTCTACTGTACTACG")
]

# 2. Crear el objeto Motif
m = motifs.create(instances)

# Imprimir información básica
print(f"Longitud del motivo: {len(m)}")
print(f"Secuencia consenso: {m.consensus}")
print(f"Secuencia consenso degenerada: {m.degenerate_consensus}")
```

### **3. Matrices asociadas al objeto Motif**

Una vez creado el objeto, puedes acceder a sus diferentes matrices asociadas:

### **A. Matriz de Frecuencias (PFM)**

Muestra el conteo directo de las bases en cada posición del alineamiento:

```python
print(m.counts)
```

### **B. Matriz de Probabilidades (PPM) e Inclusión de Pseudoconteos**

Para evitar probabilidades de cero al analizar nuevas secuencias, se aplican pseudoconteos (pseudocounts):

```python
# Convertir conteos a probabilidades ajustando el modelo de fondo (por defecto A:0.25, C:0.25, G:0.25, T:0.25)
pwm = m.counts.normalize(pseudocounts=0.5)
print(pwm)
```

### **C. Matriz de Puntuación Log-Odds (PSSM)**

Convierte las probabilidades a valores log-odds comparados con las frecuencias esperadas por azar (background):

```python
pssm = pwm.log_odds()
print(pssm)
```

### **4. Búsqueda de motivos en secuencias objetivo**

Una de las aplicaciones más comunes es buscar si un motivo PSSM aparece a lo largo de una secuencia de ADN determinada usando el método search:

```python
test_seq = Seq("GCTATACGATAACGATATCATGATTTA")

# Escanear la secuencia con el PSSM
for position, score in pssm.search(test_seq, threshold=3.0):
    print(f"Coincidencia en la posición {position} con puntuación = {score:.2f}")
```
