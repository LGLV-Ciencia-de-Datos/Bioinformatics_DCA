Ahora que hemos obtenido nuestra secuencia de proteína, vamos a realizar un BLAST (Basic Local Alignment Search Tool) del NCBI y listar los parámetros para identificar la secuencia mejor alineada.

```python
from Bio import SeqIO

protein_seq = SeqIO.read("my_sequence.fasta", "fasta")
```

```python
protein_seq.seq
```

```python
print(protein_seq.seq)
```

Aquí usaremos nuestros conocimientos de NCBI BLAST.
En esa clase usamos NCBIXML para leer el resultado de BLAST, pero aquí vamos a usar SearchIO, que eventualmente reemplazará al módulo más antiguo Bio.Blast, ya que proporciona un marco más general que también maneja otras herramientas relacionadas de búsqueda de secuencias.

```python
from Bio.Blast import NCBIWWW

# Esta celda puede tardar varios minutos (depende de los servidores del NCBI)
# blastp = BLAST de proteínas, pdb = Protein Data Bank
result_handle = NCBIWWW.qblast("blastp", "pdb", protein_seq.seq)
```

```python
from Bio import SearchIO

blast_records = SearchIO.read(result_handle, 'blast-xml')
```
